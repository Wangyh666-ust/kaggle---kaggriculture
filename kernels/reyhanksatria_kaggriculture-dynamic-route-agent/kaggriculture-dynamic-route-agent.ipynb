```python
# Validate the agent file and create the submission archive
import importlib.util
import inspect
import os
import shutil
import tarfile

if not main_path.exists():
    raise RuntimeError("Required agent file is unavailable")

current_path = Path.cwd()
os.chdir(work_path)

try:
    module_spec = importlib.util.spec_from_file_location("submission_agent", main_path)
    agent_module = importlib.util.module_from_spec(module_spec)
    module_spec.loader.exec_module(agent_module)
finally:
    os.chdir(current_path)

if not hasattr(agent_module, "agent") or not callable(agent_module.agent):
    raise RuntimeError("main.py does not expose a callable agent")

agent_signature = inspect.signature(agent_module.agent)

if list(agent_signature.parameters) != ["observation", "configuration"]:
    raise RuntimeError(f"Unexpected agent signature: {agent_signature}")

if submission_path.exists():
    submission_path.unlink()

with tarfile.open(submission_path, "w:gz") as archive:
    archive.add(main_path, arcname = "main.py")

with tarfile.open(submission_path, "r:gz") as archive:
    members = archive.getnames()

if members != ["main.py"]:
    raise RuntimeError(f"Unexpected archive structure: {members}")

main_path.unlink()

cache_path = work_path / "__pycache__"

if cache_path.exists():
    shutil.rmtree(cache_path)

print("Submission ready")
print(f"Submission path: {submission_path}")
print("Agent import: PASS")
print("Archive structure: PASS")
print("Status: READY TO SUBMIT")
```
